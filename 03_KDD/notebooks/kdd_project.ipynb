## KDD stages

**1. Selection:** Choose the anonymized European card transactions dataset, binary Class label, Time, Amount and V1..V28 variables. Document selection and sampling assumptions.

**2. Preprocessing:** Drop duplicate records; check target values; fit missing-value imputation and scaling within training folds only.

**3. Transformation:** Standardize numerical features using training-only statistics; use an 80/20 stratified holdout and seed 42.

**4. Data Mining:** Compare class-weighted logistic regression and random forest using ROC-AUC and PR-AUC.

**5. Interpretation and Evaluation:** PR-AUC is primary for rare fraud. Current experiment uses a case-enriched sample, so deployment prevalence, precision and false-alert volume need separate full-distribution evaluation.

**6. Knowledge Use:** Propose analyst triage, alert-threshold calibration, monitoring, human review and retraining; no live production deployment is claimed.

## SEMMA stages

**1. Sample:** Use the supplied retail transactions; remove rows without customer IDs, invalid timestamps, cancellations and nonpositive quantities/prices. Customer aggregation creates the analytical unit.

**2. Explore:** Examine invoices, customer frequency, skewness of spend and recency, cancellations and missing IDs. Avoid treating individual invoice lines as independent customers.

**3. Modify:** Aggregate per-customer recency, invoice frequency and monetary spend; apply log1p to skewed RFM variables and StandardScaler.

**4. Model:** Fit K-Means for k=2..6 using seed 42 and n_init=10; select k by sampled silhouette.

**5. Assess:** Compare silhouette across k and report per-cluster median RFM values. Interpret clusters without assigning unsupported business outcomes.
